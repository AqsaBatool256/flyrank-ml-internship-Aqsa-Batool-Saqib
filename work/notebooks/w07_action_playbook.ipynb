### Ranked actions + reason codes

The action queue prioritizes content using the validated model's predicted trend and the observed signals available at decision time. Each recommendation includes a reason code so a human reviewer can understand why the item was prioritized. The queue is intended for decision support and human review; it does not guarantee future content performance.


```python
# ---------------------------------------------------------
# WEEK 7 — SECTION 1
# RANKED ACTIONS + REASON CODES
# ---------------------------------------------------------

import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# ---------------------------------------------------------
# 1. LOAD THE SAME DATA USED IN WEEKS 5–6
# ---------------------------------------------------------

DATA_URL = (
    "https://raw.githubusercontent.com/"
    "AqsaBatool256/flyrank-ml-internship-Aqsa-Batool-Saqib/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(DATA_URL)

# Remove duplicate content items
df = df.drop_duplicates("content_id").copy()

# Features validated in Weeks 5–6
FEATURES = [
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position"
]

# Keep rows with complete model inputs and observed target
df = df.dropna(
    subset=FEATURES + ["trend_direction"]
).copy()

print("Rows available for action queue:", len(df))


# ---------------------------------------------------------
# 2. TRAIN THE VALIDATED WEEK-5 MODEL
# ---------------------------------------------------------

X = df[FEATURES]
y = df["trend_direction"]

model = Pipeline([
    ("scaler", StandardScaler()),
    ("logistic_regression", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

model.fit(X, y)

print("Validated model trained successfully.")


# ---------------------------------------------------------
# 3. GENERATE PREDICTED TREND + CONFIDENCE
# ---------------------------------------------------------

df["predicted_trend"] = model.predict(X)

probabilities = model.predict_proba(X)

df["prediction_confidence"] = probabilities.max(axis=1)

print("\nPredicted trend distribution:")
print(df["predicted_trend"].value_counts())


# ---------------------------------------------------------
# 4. CREATE REASON CODES
# ---------------------------------------------------------

def create_reason_code(row):

    trend = str(row["predicted_trend"]).lower()
    age = row["days_since_last_update"]
    impressions = row["impressions_90d"]
    confidence = row["prediction_confidence"]

    # Strong predicted decline + stale content
    if trend == "down" and age >= 180:
        return "DECLINING_AND_STALE"

    # Predicted decline but content is relatively recent
    elif trend == "down":
        return "PREDICTED_DECLINE"

    # Stable content with strong visibility
    elif trend == "stable" and impressions >= 500:
        return "STABLE_VISIBLE"

    # Low-confidence prediction needs human attention
    elif confidence < 0.60:
        return "LOW_CONFIDENCE"

    # Otherwise
    else:
        return "STABLE_OR_LOW_SIGNAL"


df["reason_code"] = df.apply(
    create_reason_code,
    axis=1
)


# ---------------------------------------------------------
# 5. MAP REASON CODES TO HUMAN ACTIONS
# ---------------------------------------------------------

def create_action(row):

    reason = row["reason_code"]

    if reason == "DECLINING_AND_STALE":
        return "REFRESH_REVIEW"

    elif reason == "PREDICTED_DECLINE":
        return "INVESTIGATE"

    elif reason == "LOW_CONFIDENCE":
        return "HUMAN_REVIEW"

    elif reason == "STABLE_VISIBLE":
        return "MONITOR"

    else:
        return "NO_IMMEDIATE_ACTION"


df["recommended_action"] = df.apply(
    create_action,
    axis=1
)


# ---------------------------------------------------------
# 6. RANK THE ACTION QUEUE
# ---------------------------------------------------------

# Higher priority:
# 1. Declining + stale
# 2. Predicted decline
# 3. Low confidence
# 4. Stable/visible
# 5. No immediate action

action_priority = {
    "REFRESH_REVIEW": 1,
    "INVESTIGATE": 2,
    "HUMAN_REVIEW": 3,
    "MONITOR": 4,
    "NO_IMMEDIATE_ACTION": 5
}

df["action_priority"] = df["recommended_action"].map(
    action_priority
)

# Sort by action priority and then model confidence
action_queue = (
    df.sort_values(
        by=["action_priority", "prediction_confidence"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

action_queue["rank"] = (
    action_queue.index + 1
)


# ---------------------------------------------------------
# 7. DISPLAY THE TOP 10
# ---------------------------------------------------------

queue_columns = [
    "rank",
    "content_id",
    "predicted_trend",
    "prediction_confidence",
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position",
    "recommended_action",
    "reason_code"
]

top_10_actions = action_queue[
    queue_columns
].head(10)

print("\nTOP 10 RANKED ACTIONS")
print("=====================")

display(top_10_actions)


# ---------------------------------------------------------
# 8. BASIC CHECKS
# ---------------------------------------------------------

assert len(action_queue) > 0
assert "recommended_action" in action_queue.columns
assert "reason_code" in action_queue.columns
assert "rank" in action_queue.columns

print("\nPASS — ranked action queue created.")
print("PASS — every item has a recommended action.")
print("PASS — every item has a reason code.")
```

```python
# Human-readable explanation for each recommendation

def create_reason_text(row):
    if row["reason_code"] == "DECLINING_AND_STALE":
        return (
            f"Model predicts decline; content is "
            f"{int(row['days_since_last_update'])} days old."
        )

    elif row["reason_code"] == "PREDICTED_DECLINE":
        return (
            "Model predicts a declining trend; "
            "human review is recommended."
        )

    elif row["reason_code"] == "LOW_CONFIDENCE":
        return (
            f"Prediction confidence is only "
            f"{row['prediction_confidence']:.2f}; review before acting."
        )

    elif row["reason_code"] == "STABLE_VISIBLE":
        return (
            "Content is predicted stable and has meaningful visibility; "
            "monitor rather than immediately refresh."
        )

    else:
        return "No strong action signal; continue monitoring."

action_queue["reason"] = action_queue.apply(
    create_reason_text,
    axis=1
)

# Updated top-10 queue
top_10_actions = action_queue[
    [
        "rank",
        "content_id",
        "predicted_trend",
        "prediction_confidence",
        "days_since_last_update",
        "impressions_90d",
        "recommended_action",
        "reason_code",
        "reason"
    ]
].head(10)

display(top_10_actions)

print("\nPASS — human-readable reason added to the action queue.")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

### Intended use and limits

The action queue is intended as a decision-support tool for prioritizing content that may deserve human attention. It uses observed content signals available at decision time, including days since last update, impressions over the last 90 days, CTR, and average position.

The queue can help a reviewer identify content that is predicted to be declining, especially when the content is also stale. It can support decisions about which items to investigate or consider for refresh first.

The model does not prove that a content item will decline in the future, and it does not prove that refreshing an item will improve its performance. The predictions are based on this dataset and the observed signals available during the analysis. They should not be treated as guaranteed future outcomes.

The queue is therefore intended for human-reviewed prioritization, not automatic content changes, publishing, deletion, or SEO decisions. Predictions with limited confidence should receive additional review before any action is taken.


```python
# ---------------------------------------------------------
# WEEK 7 — SECTION 2
# INTENDED USE AND LIMITS
# ---------------------------------------------------------

required_columns = [
    "content_id",
    "predicted_trend",
    "prediction_confidence",
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position",
    "recommended_action",
    "reason_code",
    "reason"
]

missing_columns = [
    col for col in required_columns
    if col not in action_queue.columns
]

print("INTENDED USE / LIMITS CHECK")
print("===========================")

print("Rows in action queue:", len(action_queue))
print("Required fields:", len(required_columns))
print("Missing fields:", missing_columns)

assert len(action_queue) > 0
assert len(missing_columns) == 0

# Confirm model inputs are observed decision-time signals
decision_time_features = [
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position"
]

print("\nDecision-time features:")
for feature in decision_time_features:
    print(" -", feature)

# Confirm target-derived fields were not used as model inputs
forbidden_features = [
    "trend_direction",
    "trend_pct"
]

print("\nLeakage-risk fields excluded from model inputs:")
for feature in forbidden_features:
    print(" -", feature)

assert all(
    feature not in FEATURES
    for feature in forbidden_features
)

print("\nPASS — action queue contains the required decision-support fields.")
print("PASS — model inputs are observed decision-time signals.")
print("PASS — target-derived fields are excluded from model inputs.")
print("PASS — predictions are framed for human-reviewed decision support.")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human review and no-go list

Every recommended action should be reviewed by a human before any content change is made. The reviewer should check the content's business purpose, search intent, current accuracy, audience needs, recent performance context, and whether the suggested action is appropriate.

A `REFRESH_REVIEW` recommendation means that the item is a candidate for investigation or refresh. It does not mean that the content must be rewritten. A `MONITOR` recommendation means that no immediate change is suggested based on the available signals.

Low-confidence predictions should receive additional scrutiny. Reviewers should also consider information that is not available to the model, such as recent campaigns, product changes, seasonality, important business events, or changes in search intent.

### No-go cases

The system should not automatically:

* publish or rewrite content;
* delete or unpublish content;
* change titles, URLs, keywords, or metadata;
* make claims about future traffic or rankings;
* declare that a refresh will improve performance;
* override editorial, legal, brand, or subject-matter review;
* make decisions using information that is not represented in the validated model.

The model is therefore a prioritization aid only. Final content decisions remain with a qualified human reviewer.


```python
# ---------------------------------------------------------
# WEEK 7 — SECTION 3
# HUMAN REVIEW + NO-GO CHECKS
# ---------------------------------------------------------

print("HUMAN REVIEW / NO-GO CHECK")
print("==========================")

# Actions that require human review
review_actions = [
    "REFRESH_REVIEW",
    "INVESTIGATE",
    "HUMAN_REVIEW"
]

# Check that these actions exist in the queue
existing_actions = set(
    action_queue["recommended_action"].dropna().unique()
)

print("Actions present in queue:")
for action in sorted(existing_actions):
    print(" -", action)

print("\nActions requiring human review:")
for action in review_actions:
    if action in existing_actions:
        print(" -", action)

# Check that the queue does not contain automatic execution commands
forbidden_automatic_actions = [
    "AUTO_PUBLISH",
    "AUTO_DELETE",
    "AUTO_REWRITE",
    "AUTO_UNPUBLISH"
]

automatic_actions_found = [
    action
    for action in forbidden_automatic_actions
    if action in existing_actions
]

print("\nForbidden automatic actions found:")
print(automatic_actions_found)

assert len(automatic_actions_found) == 0

# Confirm every recommendation has a reason
missing_reasons = action_queue[
    action_queue["reason"].isna()
]["content_id"].tolist()

print("\nItems missing human-readable reasons:", len(missing_reasons))

assert len(missing_reasons) == 0

print("\nPASS — recommendations include human-readable reasons.")
print("PASS — no automatic publishing/deletion/rewrite actions are present.")
print("PASS — final decisions remain with human reviewers.")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

### Monitoring and retrain triggers

The action queue should be monitored for changes in prediction distribution, data quality, and model usefulness. Monitoring is intended to identify when the current analysis may no longer represent the content population.

A review should be triggered if important input fields develop substantial missing values, the distribution of model predictions changes materially, or the observed outcomes begin to differ consistently from model predictions.

Retraining should not happen automatically. A new model should be considered only after a meaningful change in the data or content environment is identified and the replacement model is re-evaluated using the same leakage and grouped-validation checks.

Suggested triggers include:

* substantial change in the distribution of predicted trends;
* increased missingness in model input fields;
* changes in the content population or business context;
* repeated validation results showing declining model usefulness;
* evidence that important signals or search behavior have changed.

These are review triggers, not automatic retraining commands.


```python
# ---------------------------------------------------------
# WEEK 7 — SECTION 4
# MONITORING / RETRAIN TRIGGERS
# ---------------------------------------------------------

print("MONITORING / RETRAIN TRIGGERS")
print("============================")

# Prediction distribution
prediction_distribution = (
    action_queue["predicted_trend"]
    .value_counts(normalize=True)
    .round(4)
)

print("Current prediction distribution:")
print(prediction_distribution)

# Missingness in model inputs
missing_rates = (
    df[FEATURES]
    .isna()
    .mean()
    .round(4)
)

print("\nMissing-value rates:")
print(missing_rates)

# Basic monitoring checks
print("\nMonitoring checks:")

for feature, rate in missing_rates.items():
    print(f" - {feature}: {rate:.2%} missing")

print("\nRetrain/review triggers:")
print(" - Review if prediction distribution changes materially.")
print(" - Review if input missingness increases.")
print(" - Review if content population or search context changes.")
print(" - Re-evaluate model if validation performance declines.")
print(" - Do not retrain automatically.")

print("\nPASS — monitoring and retrain triggers documented.")
```
